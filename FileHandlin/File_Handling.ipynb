```python
with open("file.txt2","w") as f:
  f.write("this is my first line")

  # with statement ensures that the file is closed properly after the block of code is executed

  # important>> as free up system resources and avoid issues like data corruption and file lock
```

```python
with open("file.txt2","r") as f:
  line=f.readline()
  print(line)
```
