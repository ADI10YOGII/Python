```python
# regular fuunction: takes a list and gives square of each of the list

def square_number(n):

  result=[]
  for i in range(n):

    result.append(i**2)

  return result

square_number(10)
```

```python
# regular function calculates the square of the values and return in one go
# since its returning in one go>>memory usage will be more
# it will take a lot of time in execution if there is lot of computation
# as all the computations will be done first and then the calculation will be returned in one go

def square_number(n):

  result=[]
  for i in n:

    result.append(i**2)

  return result

square_number([5,10,15])
```

```python
# can we have something which generates the result one by one instead of one go
#yield: general function uses return statement but a generator function use yield ststement
```

```python
def square_number_gen(n):
  for i in range(n):
    yield i**2



square_number_gen(10) # it is an iterator object
                      # yield is creating a generator similar to iterator object
```

```python
gen=square_number_gen(10)
gen
```

```python
next(gen)
```

```python
# fibonacci series
def fib(n):
  a=0
  b=1
  for i in range(n):
    yield a
    a,b=b,a+b

```

```python
f=fib(100000000000)
f
```
