```python
# write a function that takes list as input and return all numeric values in a list
```

```python
def only_numeric(a):
  n=[]
  for i in a:
    if type(i)==int or type(i)==float:
      n.append(i)

  return n
```

```python
only_numeric([1, 2.2,"adi",True])
```

```python
only_numeric([1, 2.2,"adi",True,3+7j,"pwskills",5,6,[100,101,"yogita"]])
```

```python
def only_numeric(a):
  n=[]
  for i in a:
    if type(i)==list:
      for j in i:
         if type(j)==int or type(j)==float:
           n.append(j)


    else:
     if type(i)==int or type(i)==float:
      n.append(i)

  return n
```

```python
only_numeric([1, 2.2,"adi",True,3+7j,"pwskills",5,6,[100,101,"yogita"]])
```

```python
def marks_in_subject(**kwargs):
  marks_list=[]
  for sub,marks in kwargs.items():
    marks_list.append(marks)

  return marks_list
```
