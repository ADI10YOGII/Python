```python
class teacher_lecture:
  def lec_info(self):
    print("this is the lec info with teacher perspective ")
class student_lecture:
  def lec_info(self):
    print("this is the lec info with student perspective")
```

```python
obj1=teacher_lecture()
obj2=student_lecture()
class_obj=[obj1,obj2]
```

```python
def  parcer(class_obj):
  for i in class_obj:
    i.lec_info()# at this point ,lec_info is taking two forms wrt teacher and students
```

```python
parcer(class_obj)
```

```python
# private>> the data and method is only accessible within its class, use __ to make private
```

```python
class Student:
  def __init__(self,name,degree):
    self.name=name
    self.__degree=degree
  def show(self):
    # accessing the private data member
    print("name",self.name,'degree',self.__degree)
```

```python
# a method private
class Student:
  def __init__(self,name,degree):
    self.name=name
    self.__degree=degree
  def show(self):
    # accessing the private data member
    print("name",self.name,'degree',self.__degree)

  def __private_method(self):
    print("this is a private method")
```

```python
obj1=Student("ajay","masters")
```

```python
obj1.private_method()# throw an error
```

```python
obj1._Student__private_method()# accessing private method using class name
```

```python
# way to provide an option to see the private method>> a wrapper
class Student:
  def __init__(self,name,degree):
    self.name=name
    self.__degree=degree
  def show(self):
    # accessing the private data member
    print("name",self.name,'degree',self.__degree)

  def __private_method(self):
    print("this is a private method")

  def access_private_method(self):
    self.__private_method()
```

```python
obj1=Student("ajay","masters")
obj1.access_private_method()
```

```python
# another use case of modifying the private variable and accessing it
class Car:
  def __init__(self,model,year,speed,make):

    self.__year=year
    self.__speed=speed
    self.__make=make
    self.__model=model
  def set_speed(self,speed):
    self.__speed =0 if speed < 0 else speed
  def get_speed(self):
    return self.__speed
```

```python
c1.set_speed(-1000)
```

```python
c1.get_speed()
```

```python
c1.set_speed(1000)
```

```python
class Bank:
  def __init__(self,balance):
    self.__balance=balance
  def deposit(self,amount):
    self.__balance=self.__balance+amount
  def withdraw(self,amount):
    if self.__balance>=amount:
      self.__balance=self.__balance-amount
      return True
    else:
      return False
  def get_balance(self):
    return self.__balance
```

```python
acc1=Bank(1000)
```

```python
acc1.get_balance()
```

```python
acc1.deposit(500)
```

```python
acc1.get_balance()
```

```python
acc1.withdraw(100)
```

```python
acc1.get_balance()
```

```python
acc1.withdraw(1500)
```

```python
# protected>> within the class and its subclass,protected member can be access,(_)
```

```python
class College:
  def __init__(self):
    self._college_name="PWSkills"
class Student(College):
  def __init__(self,name):
    self.name=name
    College.__init__(self)# accessing variable of base class>>classname__init__(self)
  def show(self):
    print("name",self.name,"college",self._college_name)# directly call the base class variable using one _
```

```python
stud=Student("Ajay")
```

```python
stud.name
```

```python
stud.show()
```

```python
coll=College()
coll._college_name # accessing the protected variables
```

```python
# another way to access the data of base class using super() class
class College:
  def __init__(self):
    self._college_name="PWSkills"
class Student(College):
  def __init__(self,name):
    self.name=name
    super().__init__()# include super and remove self,base class is super class
  def show(self):
    print("name",self.name,"college",self._college_name)# directly call the base class variable using one _
```

```python
stud=Student("Ajay")
```

```python
stud.show()
```
