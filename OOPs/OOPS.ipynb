```python
# we want to take the attribute value of the class for each object differently
class ListOps:
  def __init__(self,l):# initialization of variable
    self.l=l

  def extract_even(self,l):
    l1=[]
    for i in self.l:
      if i%2==0:
        l1.append(i)
    return l1
  def extract_even1(self,l):
    l1=[]
    for i in l:
      if i%2==0:
        l1.append(i)
    return l1
  def extract_odd(self,l):
      l1=[]
      for i in self.l:
        if i%2!=0:
          l1.append(i)
      return l1
```

```python
class ListOps:
  def __init__(self,l):# initialization of variable
    self.l2=l # self is a reference to the class ,class knows i have a variable l
              # everyehere you will be able to access l using self.l2
  def extract_even(self):     # init has been used only pass self
      l=self.l2              # to use l just make l=self.l
      l1=[]
      for i in l:
       if i%2==0:
         l1.append(i)
      return l1


  def extract_odd(self):
      l=self.l2
      l1=[]
      for i in l:
        if i%2!=0:
          l1.append(i)
      return l1
```

```python
ops1=ListOps([1,8,9,11,12])
```

```python
ops1.extract_even()
```

```python
ops1.extract_odd()
```

```python
# multiple attributes
class Book:
   def __init__(self,name,author,title):
     self.name_0f_book = name
     self.book_author = author
     self.title_name = title

   def extract_details_name_title(self):
    print(self.name_0f_book,self.title_name)

   def extract_details_name_author(self):
    print(self.name_0f_book,self.book_author)
```

```python
student1.extract_details_name_author()
```
