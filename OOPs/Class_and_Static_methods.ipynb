```python
# class methods>> bound to the classs(and not to the particular instance of the class)
# can access /modify methods and attributes associated to class across all the instances
# @ classmethod decorator
# insted of self,cls as first parameter
# alternative of init method to pass the data
```

```python
class Student:
  # constructor>> constructing class with initialization of variables
  def __init__(self,name):
    self.name=name
```

```python
obj=Student("Ajay")
```

```python
obj.name
```

```python
class Student:

  def __init__(self,name):
    self.name=name# self means this variable /method is referring to current instance of the class

  @classmethod # bounds/binds to the class
  def student_details(cls,name): # cls instead of self as a first parameter
                    # this method is associated/referring to the class itself
    return cls(name)
```

```python
# what is meaning that it  is referring the class
# earlier to access any method/attribute >>first make instance /object of class
# then obj.attribute
# but if it is class method,you can directly access it
```

```python
Student.student_details("ajay")
```

```python
obj1=Student.student_details("ajay")# since students_details directly binds to the class
# you can directly call it without making any object
```

```python
obj1.name
```

```python
# seeing class method as alternative of init method
# actually class method is overloading the init method
```

```python
#class method is overloading the init method
# modifying the class/instance method
class Student:

  def __init__(self,name):# init method is an instance method
    self.name=name

  @classmethod
  def student_details(cls,name1):
    return cls(name1)
```

```python
stud=Student.student_details("ajay")
```

```python
stud.name # you will get ajay inspite the variable name1
# because it is overloading the init method
```

```python
# class method can modify class state applicable to all the instances of the class
class Student:

  total_students=0 # class variable

  def __init__(self,name):
    self.name=name # instance variable
    Student.total_students+=1 # total_student is the class variable associated associated with all the instances

  @classmethod
  def get_total_students(cls):
    return cls.total_students
```

```python
Student.total_students
```

```python
std1=Student("ajay")
std2=Student("bijay")
```

```python
std1.name
```

```python
std2.name
```

```python
Student.total_students
```

```python
Student.get_total_students() # class method can modify class state applicable to all the instances of the class
# that is why you see the increase in the count
```

```python
# to make external function as class method
class Student:

  total_students=0 # class variable

  def __init__(self,name):
    self.name=name # instance variable
    Student.total_students+=1 # total_student is the class variable associated associated with all the instances

  @classmethod
  def get_total_students(cls):
    return cls.total_students
```

```python
def course_details(cls,course_name):
  print("the details of the course are:",course_name)
```

```python
Student.course_details=classmethod(course_details)
# adding external function as class method
```

```python
Student.course_details("Data science")
```

```python
# delete any class method
del Student.course_details
```

```python
Student.course_details
```

```python
Student.get_total_students()
```

```python
# another way to delete
delattr(Student,"get_total_students")
```

```python
Student.get_total_students()
```

```python
# static method
# does not modify class or instance state
# it behaves like a plain function that belongs to class
# it dosen't take either self or cls
# dosent access instance or class state
# dosent depend on class/instance
# works with the parameter passed
```

```python
class Calculator:
  @staticmethod
  def add(x,y):
    return x+y
  @staticmethod
  def subtract(x,y):
    return x-y
```

```python
Calculator.add(10,20)
```

```python
Calculator.subtract(10,20)
```
