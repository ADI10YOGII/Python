```python
class MyClass:
  def __init__(self,x):
    self.x=x
  def __repr__(self):
    return f"MyClass({self.x})"
```

```python
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __eq__(self,other):
    return self.x==other.x and self.y==other.y
```

```python
p1=Point(1,2)
p2=Point(1,2)
print(p1==p2) # comparing both of the objects
#when you are using == operator with point Objects,python internally calls __eq__
```

```python
class Point:
  def __init__(self,x,y):
    self.x=x
    self.y=y
  def __add__(self,other):
    return Point(self.x+other.x , self.y+other.y)
```

```python
p1=Point(1,2)
p2=Point(1,2)
p3=p1+p2 # when you are using + operator with point Objects,python internally calls __add__
print(p3.x,p3.y)
```
