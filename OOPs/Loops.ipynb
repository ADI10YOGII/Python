```python
# while loop: repeatedly execute a block of code until a condition is met
```

```python
n=7
i=1
while i<n:
  print(i)
  i=i+1
```

```python
n=7
i=1
while i<n:
  print(i)
  i=i+1
else:
  print("this will be executed when while loop runs successfully without any break")
```

```python
# break: terminates or exits the loop(while else is a loop)
n=7
i=1
while i<n:
  print(i)
  i=i+1
  if i==3:
    break
else:
  print("this will be executed when while loop runs successfully without any break")
```

```python
# continue: skips the rest iteration, go back to the loop
n=7
i=1
while i<n:
  i=i+1
  if i==3:
    continue
  print(i)
else:
  print("this will be executed when while loop runs successfully without any break")
```

```python
#print a right triangle with *  :using while loop


row=1
while row<=4:
  col=1
  while(col<=row):
    print('*',end="")

    col=col+1

  print()
  row=row+1




```

```python
#print a right triangle with *  :using for loop

for i in range(4):
  for j in range(i+1):
    print('*',end=" ")

  print()
```
