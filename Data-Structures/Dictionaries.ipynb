```python
#zip function

students=["Arun","Ajay","Bijay"]
marks=[80,95,100]
for i in zip(students,marks):
  print(i)
```

```python
type(i)
```

```python
# store the above info in key value pair
```

```python
students=["Arun","Ajay","Bijay"]
marks = [80,95,100]
students_marks = {}
for student , mark in zip(students,marks):
  students_marks[student]= mark


```

```python
students_marks
```

```python
# using dictionary comprehension method
students=["Arun","Ajay","Bijay"]
marks = [80,95,100]
{student:mark for student,mark in zip(students,marks)}
```

```python
students=["Arun","Ajay","Bijay"]
marks = [80,95,100]
{student:marks for student,marks in zip(students,marks)}
```
