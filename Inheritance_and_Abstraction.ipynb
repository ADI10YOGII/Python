```python
# inheritance :process of child receiving the properties of parent class
# syntax
# class BaseClass:
#    body of base class
#class DerivedClass(BaseClass):
#     body of derived class
```

```python
# Single inheritance>> when a derived class has only one parent class
```

```python
class Father:
  def father_property(self):
    print("this is the father property")
class Child(Father):
  def child_property(self):
    print("this is the child property")
```

```python
# child can have his property and also his father property
child_obj=Child()
```

```python
child_obj.child_property()
```

```python
child_obj.father_property()
```

```python
# father will have only his property
```

```python
# method overriding: redefining methods of parent class in child class
# always method of child class will be executed
# happens between two classes
class Fruit:
  def fruit_info(self):
    print("inside parent class")

class Apple(Fruit):
  def fruit_info(self):
    print("inside child class")

  def apple_info(self):
    print("inside the child class")
```

```python
obj=Apple()
obj.apple_info()
```

```python
obj.fruit_info()
```

```python
# Multi level inheritance
```

```python
class GrandFather:
  def grand_father_property(self):
    print("this is the grand father property")
class Father(GrandFather):
  def father_prop(self):
    print("father property as well as grand father property")
class Son(Father):
  def son_prop(self):
    print("i have all the property")
```

```python
son=Son()
son.son_prop()
```

```python
son.grand_father_property()
```

```python
fath=Father()
fath.grand_father_property()
```

```python
# multiple inheritance: one child class inherit the property of multiple parent class
```

```python
class ParentClass1:
  def method1(self):
    print("this is the method of parent class 1")
class ParentClass2:
  def method2(self):
    print("this is the method of parent class 2")
```

```python
class childClass(ParentClass1,ParentClass2):
  def method3(self):
    print("this is the method of child class")
```

```python
c1=childClass()
c1.method3()
```

```python
c1.method2()
```

```python
c1.method1()
```

```python
# diamond problem >> it occurs when a class inherits from 2 or more than 2 class
# will lead to ambiguity in execution of methods
# to remove diamond problem :python uses method resolution order(MRO) algorithm called c3 linearization
# means the class that is inherited first in the derived class,that method will be executed
# method of parent class1 will be executed
class ParentClass1:
  def method_par(self):
    print("this is the method of parent class 1")
class ParentClass2:
  def method_par(self):
    print("this is the method of parent class 2")
class childClass(ParentClass1,ParentClass2):
  def method3(self):
    print("this is the method of child class")
```

```python
c2=childClass()
c2.method_par()
```

```python
# diamond problem
class A:
  def method(self):
    print("this is the method of class A")
class B(A):
  def method(self):
    print("this is the method of class B")
class C(A):
  def method(self):
    print("this is the method of class C")
class D(C,B):
  pass
```

```python
d=D()
d.method()
```

```python
# Hybrid inheritance : multiple types or combination of diffrent inheritance

class Vehicle:
  def info(self):
    print("inside vehicle class")
class Car(Vehicle):
  def car_info(self):
    print("inside car class")
class Truck(Vehicle):
  def truck_info(self,name):
    print("This is truck info",name)

class SportsCar(Car,Vehicle):
  def sports_car_info(self):
    print("inside sports car class")
```

```python
#Abstraction: to hide the complex implementation
# hiding the unneeded details and exposing the required
# Abstract class cannot be instantiated(can not make object)
# abstract class should always be subclassed
# abstract class may contain methods that are just declared and not implemented in abstract class itself,subclass are responsible for implementing these abstract methods
#if in inherited class a method is not present from abstract class,it will not throw the error
```

```python
# abstraction can be achieved using abstract class>>import abc
```

```python
import abc # collection of multiple functions is a module
# in any class at pwskills >>data science,web d,the following three methods will be same
class PwSkills:

  @abc.abstractmethod
  def student_details(self):
    pass
  @abc.abstractmethod
  def student_assignment(self):
    pass
  @abc.abstractmethod
  def student_marks(self):
    pass
```

```python
class DataScience(PwSkills):
  def student_details(self):
    return "Data Science course details"
  def student_marks(self):
    return "give ds student marks"
```

```python
class WebDev(PwSkills):
  def student_assignment(self):
    return "this will give web dev student assignment"
  def student_marks(self):
    return "webdev student marks"
```

```python
ds=DataScience()
```

```python
ds.student_assignment()# it will not throw any error even if any method is not defined
```

```python
ds.student_details()
```

```python
w=WebDev()
w.student_assignment()
```

```python
class Shape:
  @abc.abstractmethod
  def calculate_area(self):
    pass
class Rectangle(Shape):
  def calculate_area(self):
    return "Area of rectangle is len*breadth"
class Circle(Shape):
  def calculate_area(self):
    return "Area of circle is pi*r*r"
```

```python
rect=Rectangle()
rect.calculate_area()
```

```python
c=Circle()
c.calculate_area()
```
