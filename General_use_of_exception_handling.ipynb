```python
try:
  10/0 # suspicious

except ZeroDivisionError as e:
  print("Here i am handling the zero division error",e)

  ##This block will only trigger if the error is exactly a division by zero (like 10/0).
##If any other unexpected error occurs inside that try block (for example, a NameError or a TypeError like 10/"string"), this block will not catch it.
```

```python
try:
  10/"str" # suspicious

except ZeroDivisionError as e:
  print("Here i am handling the zero division error",e)
```

```python
# best practise :always catch specific error
# specifity in exception handling
```

```python
# another wxample
try:
  10/"some string"

except TypeError as e:
  print("Some string is being used in division ",e)
```

```python
# benefit of handling specific error>> you know what error is expected and if
# another errror comes which you are not able to think,you will be able to address that
```

```python
# another wxample
try:
  10/"3"

except TypeError as e:
  print("Some string is being used in division ",e)
  print("The calculation is",10/int("3"))
```

```python
try:
  int("pwskills")
except ValueError as e:
  print("The string can not be converted to integer",e)
```

```python
# handling two specific exceptions/multiple exceptions
try:
  10/"3"

except (ZeroDivisionError,TypeError) as e:
  print("The division is not possible due to the error",e)
```

```python
try:
  10/0

except (ZeroDivisionError,TypeError) as e:
  print("The division is not possible due to the error",e)
```

```python
# handling multiple exceptions in separate blocks
try:
  10/0

except ZeroDivisionError as e:
  print("The division is not possible due to the error",e)

except ValueError as e:
  print("This is value error",e)
```

```python
try:
  10/"string"

except ZeroDivisionError as e:
  print("The division is not possible due to the error",e)

except TypeError as e:
  print("This is type error",e)
```

```python
# sometime you will not know what could be the possible exceptions

try:
  f=open("test.txt","r")
  f.read()
except Exception as e: # parent class
    print("File is not opened due to",e)
except FileNotFoundError as e: ## unreachable
  print("The file is not found",e)
```

```python
# in case you know specific exception keep that block first and block of generic exception after that
try:
  f=open("test.txt","r")
  f.read()
except FileNotFoundError as e:
  print("The file is not found",e)
except Exception as e: # parent class
    print("File is not opened due to",e)
```

```python
# best practise >> always try to log
import logging
logging.basicConfig(filename="program.log",level=logging.DEBUG)
try:
  f=open("example.txt","r")
  f.read()
except FileNotFoundError as e:
  logging.error(f"The file is not found{e}")
```

```python
#putting exception handling in a function
def test(file):
  try:
    f=open(file,"r")
    f.read()
  except FileNotFoundError as e:
    print(f"The file is not found{e}")
```

```python
test("example.txt")
```
