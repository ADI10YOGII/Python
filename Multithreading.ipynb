```python
import time
import threading # python module
start=time.perf_counter()

def test_func():
  print("do something")
  print("sleep for 1 sec")
  time.sleep(1)
  print("done with sleeping")
# run the program on two threads

t1=threading.Thread(target=test_func)# Threading module import thread
# test_func we are only passing it not calling
t2=threading.Thread(target=test_func)

t1.start() # to start the thread
t2.start()



end=time.perf_counter()


print(f"The program finished in {round(end-start,2)} seconds")
```

```python
import time
import threading # python module
start=time.perf_counter()

def test_func():
  print("do something")
  print("sleep for 1 sec")
  time.sleep(1)

  print("done with sleeping")
# run the program on two threads

t1=threading.Thread(target=test_func)# Threading module import thread
# test_func we are only passing it not calling
t2=threading.Thread(target=test_func)

t1.start() # to start the thread
t2.start()

t1.join()# join first executed these t1,t2 threads and then the main thread will be executed
t2.join()


end=time.perf_counter()


print(f"The program finished in {round(end-start,2)} seconds")
```

```python
import time
import threading # python module
start=time.perf_counter()

def test_func():
  print("do something")
  print("sleep for 1 sec")
  time.sleep(1)

  print("done with sleeping")
# run the program on two threads


threads=[]
for i in range(10):
  t=threading.Thread(target=test_func)
  t.start()
  threads.append(t)

for thread in threads:
  thread.join()




end=time.perf_counter()


print(f"The program finished in {round(end-start,2)} seconds")
```

```python
# since test_func is called 10 times,it should have have taken 10 secs but due to multithreading got completed in 1 sec
```

```python
import time
import threading # python module
start=time.perf_counter()

def test_func(args):
  print("do something")
  print(f"sleep for{args} sec")
  time.sleep(args)

  print("done with sleeping")
# run the program on two threads


threads=[]
for i in range(10):
  t=threading.Thread(target=test_func,args=[2])
  t.start()
  threads.append(t)

for thread in threads:
  thread.join()




end=time.perf_counter()


print(f"The program finished in {round(end-start,2)} seconds")
```

```python
# use case :
# multithreading works well with I/O bound task
# means where some output has to wait for input
# reading , writing files , network communication,data base queries
```
