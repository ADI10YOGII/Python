```python
# factorial

def fact_num(n):
  if n==0:
    return 1
  else:
    return n*fact_num(n-1)
```

```python
fact_num(5)
```

```python
fact=lambda n: 1 if n==0 else n*fact(n-1)
fact(5)
```

```python
# reduce>> folding/reduction
#syntax>> reduce(func,iterable)
# reduce will always take 2 argument function

from functools import reduce
```

```python
l=[2,1,3,4,4,5,6]
reduce(lambda x,y:x+y,l)
```

```python
l=[2,1,3,4,5,6]
reduce(lambda x,y,z:x+y+z,l)
```

```python
reduce(lambda x,y:x*y,l)
```

```python
words=["Data","Science","Courses"]
reduce(lambda x,y:x+" "+y,words)
```

```python
numbers=[1,2,5,3,100,1000,6,100000]
reduce(lambda x,y:x if x>y else y,numbers)
```

```python
def factorial(n):
  return reduce(lambda x,y:x*y,range(1,n+1))
```

```python
factorial(5)
```

```python
#filter function:it is used to filter elements from an iterable based on some condition
#syntax>>filter(function,iterable)
```

```python
def check_even(num):
  if num%2==0:
    return True
  else:
    return False
```

```python
num=[1,2,4,6,8,9,10]
list(filter(check_even,num))
```
