```python
import csv
with open("example_csv.csv","w") as f:  # f is a file object
  w=csv.writer(f) # writer object right now empty file object
  #Creates a CSV writer object (w) associated with the opened file f. This object converts your data into CSV-formatted strings.
  for i in data:
    w.writerow(i)
```

```python
with open("example_csv.csv","r") as f:
  r=csv.reader(f)
  for i in r:
    print(i)
```

```python
# writing a binary data in a file
# wb: write byte mode
with open("test_bin.bin","wb") as f:
  f.write(b"\x48\x65\x6c\x6c\x6f\x2c\x20\x57\x6f\x72\x6c\x64\x21")
#  \x48\x65\x6c\x6c\x6f\x2c\x20\x57\x6f\x72\x6c\x64\x21 :(byte form) represents Hello World!
# open in notebook
```

```python
with open("test_bin.bin","rb") as f:
  print(f.read())
```

```python
# Buffered reading and writing
# it allows the program to read and write data in chunks to avoid crashing of machine say 100gb file
```

```python
import io # io: means input and output
with open("test_buf.txt","wb") as f:
  file=io.BufferedWriter(f)
  file.write(b"A computer is a machine that can be programmed to automatically carry out sequences of arithmetic or logical operations (computation). Modern digital electronic computers can perform generic sets of operations known as programs, which enable computers to perform a wide range of tasks. The term computer system may refer to a nominally complete computer that includes the hardware, operating system, software, and peripheral equipment needed and used for full operation; or to a group of computers that are linked and function together, such as a computer network or computer cluster.")
  file.write(b"Hello world,how are you?")
  file.write(b"I am studying ds")
  file.flush() # writing the files in chunks
```

```python
with open("test_buf.txt","rb") as f:
  file=io.BufferedReader(f)
  data=file.read(100)# read the first chunk of 100 bytes
  print(data)
```
