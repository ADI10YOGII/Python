```python
# from print statement you understand that when b is 0,it throws the error
def divide(a,b):
  print(a,b)
  if b==0: # breaking of code in terms of execution
    return "zero division error"
  return a/b

for i in range(-10,10):

  divide(i,i+1)
```

```python
# print statement is not a pythonic way to debug >>print statements are not efficient
# print statement is good for simple scripts but not complex
# in console/terminals all the print statements will be lost
# you use logging instead of print statement


#logging >> it reccords the state and flow of your program/code/software
# it is useful for understanding,monitoring and debugging of your code
# it shows how program behaves over time


#Analogy >> diary entry since childhood to now>> you will understand how you evolve
#similarly in complex scripts to understand how your code is changing the result over time
# you can log the specific steps
```

```python
# DEBUG>> lowest level of logging .To give detailed information of any variable , msg,while doing debugging
# INFO>> this is used to convey that the code is working as expected
#example>> "data analysis is completed"
# WARNING >> used to indicate that something  unexpected happened or ppotential issue in the code
# ERROR >> serious problem with some function
# CRITICAL >> Termination of program/code/software


# the above level of logging is executed in the same order>> debug being lowest priorit
# and critical being highest priority

# while defining level in logging ,basicConfig, from that level the logs will be logged or in other words the log priority will start from that level which is used to initialize the loggin
# if you initialize with Warning then only warning ,error,crirical msgs will be logged
# and info will be logged
```

```python
# write a program to separate the integer and string in two lists separately
l=[1,"hello",[2,"world"],4,[3,"python"]]
```

```python
l
```

```python
l1_int=[]
l2_str=[]

for i in l:
  logging.info(f"processing each element{i}")


  if type(i)==list:

    for j in i:
      logging.info(f"processing each element{j}")

      if type(j)==int:
        l1_int.append(j)

  elif type(i)==int:
    l1_int.append(i)

  else:
    l2_str.append(i)

logging.info(f"The result is : {l1_int} ,{l2_str}")
logging.shutdown()
```
