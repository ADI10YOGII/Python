```python
#syntax of try except
#try:
        #suspicious code
#except:
        # executed when exception occurs in try block

try:
    f=open("example.txt","r")
except Exception as e:
  print("There is an issue with the code and the issue is:",e)
print(6+3)
print("This is my next step")
```

```python
# finally block >> always executed
# even if you enter exception block or not, finally will always be executed
# similar to running other lines of code
try:
    f=open("example.txt","w")
    f.write("Hello to exceptional Handling")# you will not be able to see written content until you close the file
    # close the file in else block
except Exception as e:
  print("There is an issue with the code and the issue is:",e)
else:
  print(5+3)
  print("This block will be executed iif try except block is executed without any exception")
finally:
  print("This will be executed always")
```

```python
#custom exception class

class ValidateSalary(Exception): # Exception is the parent class where all types of python exceptions are present

  def __init__(self,msg):
    self.msg=msg
```

```python
def validate_salary(salary):
  if salary<=0:
    raise ValidateSalary("Salary cannot be negative")

  elif salary >= 300000000:
    raise ValidateSalary("Salary is unexpected")
  else:
   print("salary is valid")
```

```python
try:
  salary=int(input("Enter your salary"))
  validate_salary(salary)
except ValidateSalary  as e:
  print(e)
```
